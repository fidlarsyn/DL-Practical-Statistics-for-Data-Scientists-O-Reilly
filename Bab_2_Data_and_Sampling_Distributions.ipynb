### Bab ini membahas konsep fondasi pengambilan sampel (*sampling*) dan distribusi inferensial[2][8]. Di era *big data*, prinsip sampling acak tetap krusial untuk mengurangi bias, menghemat waktu komputasi, dan mengevaluasi tingkat ketidakpastian mode

# **1\. Random Sampling and Sample Bias**
###Sampel (*sample*) adalah subset yang diambil dari dataset yang lebih besar yang disebut populasi (*population*). Pengambilan sampel acak (*random sampling*) memberikan setiap anggota populasi peluang yang sama untuk terpilih, menghasilkan *simple random sample*. Sampling dapat dilakukan dengan pengembalian (*with replacement*) atau tanpa pengembalian (*without replacement*). Dalam *data science*, representativitas sampel lebih penting daripada kuantitas data mentah.

# **2. Bias**
### Bias statistik (*statistical bias*) merujuk pada kesalahan sistematis yang disebabkan oleh proses pengukuran atau sampel, bukan variasi acak (*random error*). Proses yang tidak bias (*unbiased*) mungkin tetap memiliki eror acak, namun eror tersebut berpusat di sekitar nilai sebenarnya dan tidak cenderung mengarah ke satu arah tertentu.

# **3. Random Selection**
### Pemilihan acak adalah cara terbaik untuk menghindari *sample bias* dan memastikan keterwakilan populasi. Tantangan utamanya terletak pada mendefinisikan populasi sasaran secara presisi (misalnya kriteria spesifik observasi apa yang tergolong sebagai "pelanggan").

# **4. Size Versus Quality: When Does Size Matter?**
### Kualitas data (kebersihan, kelengkapan, dan keterwakilan) jauh lebih utama dibandingkan kuantitas data. Dataset berukuran sangat besar (*big data*) baru benar-benar dibutuhkan ketika menganalisis kejadian langka (*rare events / long tail*).

# **5. Sample Mean Versus Population Mean**
### Untuk membedakan statistik sampel dengan parameter populasi, statistik menggunakan simbol yang berbeda. Rata-rata sampel disimbolkan dengan $\bar{x}$ (x-bar), sedangkan rata-rata populasi disimbolkan dengan $\mu$ (mu). Nilai populasi umumnya diinferensi berdasarkan sampel yang diamati.

# **6. Selection Bias**
### *Selection bias* terjadi akibat prosedur seleksi data yang secara tidak sadar memicu hasil yang keliru (misalnya *self-selection bias* atau *data snooping*). Menguji terlalu banyak hipotesis atau mengulang-ulang pencarian model pada data yang sama (*vast search effect*) juga berisiko menghasilkan pola semu.

# **7. Regression to the Mean**
### Fenomena di mana pengukuran ekstrem pada satu kesempatan cenderung diikuti oleh pengukuran yang lebih mendekati nilai rata-rata pada kesempatan berikutnya. Fenomena ini terjadi akibat variasi acak, bukan karena perubahan kausalitas nyata

# **8. Sampling Distribution of a Statistic**
### Sampling distribution adalah distribusi frekuensi dari suatu statistik sampel (seperti rata-rata sampel) yang dihitung dari banyak sampel acak yang ditarik dari populasi yang sama. Semakin besar ukuran sampel $n$, distribusi dari rata-rata sampel akan semakin menyempit di sekitar nilai rata-rata populasi.

**Code Reproduction: R:**

```python
library(ggplot2)
# 1. Sampel data individu
samp_data <- data.frame(income=sample(loans_income, 1000), type='data_dist')$0
# 2. Sampel rata-rata n=5
samp_mean_05 <- data.frame(
  income = tapply(sample(loans_income, 1000*5), rep(1:1000, rep(5, 1000)), FUN=mean),
  type = 'mean_of_5')
    type = 'mean_of_5')
# 3. Sampel rata-rata n=20
samp_mean_20 <- data.frame(
  income = tapply(sample(loans_income, 1000*20), rep(1:1000, rep(20, 1000)), FUN=mean),
  type = 'mean_of_20')
# Plot histogram perbandingan
income <- rbind(samp_data, samp_mean_05, samp_mean_20)
income\$type = factor(income\$type, levels=c('data_dist', 'mean_of_5', 'mean_of_20'),
                      labels=c('Data', 'Mean of 5', 'Mean of 20'))
```

# **9. Central Limit Theorem**
### Teorema Limit Terpusat (Central Limit Theorem / CLT) menyatakan bahwa distribusi dari statistik rata-rata sampel akan mendekati bentuk kurva lonceng (distribusi normal) seiring bertambahnya ukuran sampel n, meskipun distribusi populasi aslinya tidak normal.

# **10. Standard Error**
### Standard Error (SE) mengukur variabilitas statistik sampel antar sampel yang berbeda1820. Dirumuskan sebagai $SE = s / \sqrt{n}$, di mana $s$ adalah standar deviasi sampel dan $n$ adalah ukuran sampel20. Standard Error mengukur penyebaran statistik sampel, sedangkan Standard Deviation mengukur penyebaran nilai data individu.

# **14. Normal Distribution**
### Kurva lonceng simetris tradisional di mana 68% data berada dalam 1 standar deviasi dari rata-rata ($\mu \pm 1\sigma$) dan 95% data berada dalam 2 standar deviasi ($\mu \pm 2\sigma$)31. Mayoritas data mentah tidak berdistribusi normal, namun distribusi sampling dari statistik sampel sering kali mendekati normal.


# **15. Standard Normal and QQ-Plots**
### Standard Normal Distribution (z-distribution) adalah distribusi normal dengan mean = 0 dan standar deviasi = 13133. Data ditransformasikan menjadi z-scores ($z = (x - \bar{x})/s$)3233. QQ-Plot (Quantile-Quantile Plot) memplot z-score sampel terhadap kuantil normal teoritis untuk melihat seberapa dekat data sampel dengan distribusi normal.

**Code Reproduction: R:**

```python
norm_samp <- rnorm(100)
qqnorm(norm_samp)
abline(a=0, b=1, col='grey')$0
```

# **18. Binomial Distribution**
### Distribusi frekuensi dari jumlah keberhasilan ($x$) dalam $n$ kali percobaan independen (Bernoulli trials) dengan peluang sukses $p$ pada setiap percobaan. Digunakan untuk memodelkan hasil biner (ya/tidak, sukses/gagal). Rata-ratanya adalah $n \times p$ dan variansinya $n \times p \times (1-p).

**Code Reproduction: R:**

```python
# Probabilitas tepat 2 sukses dalam 5 percobaan (p=0.1)
dbinom(x=2, size=5, p=0.1)
# Probabilitas 2 atau kurang sukses dalam 5 percobaan
pbinom(2, 5, 0.1)$0
```
