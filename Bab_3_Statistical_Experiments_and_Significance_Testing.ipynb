# **Ringkasan**
### Desain eksperimen adalah fondasi utama pengujian statistik untuk mengonfirmasi atau menolak suatu hipotesis. Di dunia *data science*, eksperimen dilakukan secara terus menerus (seperti uji tampilan antarmuka web dan strategi pemasaran) untuk menguji efektivitas perlakuan (*treatment*) baru dibandingkan dengan kondisi standar.


# **1. A/B Testing**

### Uji A/B (*A/B Test*) adalah eksperimen yang membandingkan dua kelompok (*treatment A vs treatment B*) untuk menentukan perlakuan mana yang memberikan hasil lebih baik. Biasanya, salah satu kelompok berfungsi sebagai **Control Group** (perlakuan standar/tanpa perlakuan) dan kelompok lainnya sebagai **Treatment Group**. Subjek (misal: pengunjung web) dialokasikan secara acak (*randomization*) ke dalam kelompok A atau B, dan efeknya diukur menggunakan instrumen statistik tunggal (*test statistic*).

# **2. Why Have a Control Group?**
###Tanpa kelompok kontrol (*control group*), tidak ada jaminan bahwa perubahan hasil benar-benar dipicu oleh perlakuan (*treatment*) dan bukan oleh faktor ekstrim/luar lainnya (seperti perubahan musim, waktu, atau karakteristik pengguna). Kelompok kontrol memastikan kondisi *ceteris paribus* (semua hal lain dianggap sama).

# **3. Why Just A/B? Why Not C, D,...?**

###Eksperimen web tidak terbatas pada dua opsi. Peneliti dapat menguji banyak variasi sekaligus (A/B/C/D). Namun, semakin banyak variasi yang diuji secara bersamaan dalam statistik tradisional, semakin rumit pengujian signifikansinya. Dalam *data science*, fokus utama sering kali beralih dari sekadar menjawab "apakah A berbeda dari B?" menjadi "mana opsi terbaik dari banyak pilihan?" yang nantinya diselesaikan dengan algoritma *Multi-Arm Bandit*.

# **4. Hypothesis Tests**

### Pengujian hipotesis (*hypothesis testing*) diciptakan untuk melindungi peneliti agar tidak terkecoh oleh variasi acak (*random chance*). Otak manusia cenderung mencari pola dan menganggap perbedaan acak kecil sebagai sesuatu yang signifikan secara nyata.

# **5. The Null HypothesisRingkasan:
### Hipotesis Nol ($H_0$) adalah andaikan dasar bahwa tidak ada perbedaan nyata antara kelompok A dan B, serta perbedaan yang teramati semata-mata disebabkan oleh kebetulan/variasi acak. Tujuan eksperimen adalah mengumpulkan bukti untuk menolak atau meragukan Hipotesis Nol ini.**

# **6. Alternative HypothesisRingkasan:**
### Hipotesis Alternatif ($H_a$) adalah lawan dari Hipotesis Nol. Jika $H_0$ menyatakan "A = B", maka $H_a$ menyatakan "A ≠ B" (atau "B lebih baik dari A").

# **7. One-Way Versus Two-Way Hypothesis Tests**

### Pengujian satu arah (*one-tail / one-way test*) hanya memperhitungkan arah perbedaan yang diharapkan (misal: B &gt; A). Pengujian dua arah (*two-tail / two-way test*) lebih konservatif karena memperhitungkan kemungkinan perbedaan ekstrem di kedua arah (A &gt; B atau B &gt; A).

# **8. Resampling**
### *Resampling* adalah proses pengambilan sampel berulang dari data hasil observasi. Dua teknik utama *resampling* adalah *bootstrapping* (untuk mengukur presisi estimasi) dan *permutation test* (untuk menguji hipotesis antar kelompok).

# **9. Permutation Test**

### Dalam uji permutasi (*permutation test* atau *randomization test*), data dari seluruh kelompok digabungkan menjadi satu *pool* (mencerminkan Hipotesis Nol bahwa semua perlakuan sama), kemudian dikocok (*shuffled*) dan dibagi kembali secara acak menjadi kelompok-kelompok baru dengan ukuran yang sama dengan sampel asli. Selisih antar kelompok acak ini dihitung berulang kali untuk membentuk distribusi acak Hipotesis Nol.

# **10. Example: Web StickinessRingkasan:**
### Studi kasus membandingkan waktu sesi (session time) pengguna pada Halaman A (21 sampel) dan Halaman B (15 sampel). Rata-rata waktu sesi Halaman B lebih lama 35.67 detik dibanding Halaman A. Uji permutasi dijalankan sebanyak 1.000 kali untuk melihat seberapa sering kocokan acak menghasilkan perbedaan $\ge 35.67$ detik.

**Code Repodruction: R:**

```python
library(ggplot2)

# 1. Boxplot Visualisasi
ggplot(session_times, aes(x=Page, y=Time)) + geom_boxplot()
# 2. Perhitungan Selisih Mean Observasi
mean_a <- mean(session_times[session_times['Page'] == 'Page A', 'Time'])
mean_b <- mean(session_times[session_times['Page'] == 'Page B', 'Time'])
obs_diff <- mean_b - mean_a
# 3. Fungsi Permutasi Kustom
perm_fun <- function(x, nA, nB) {
  n <- nA + nB
  idx_b <- sample(1:n, nB)
  idx_a <- setdiff(1:n, idx_b)
  return(mean(x[idx_b]) - mean(x[idx_a]))
\}
# 4. Simulasi 1.000 kali
perm_diffs <- rep(0, 1000)
for (i in 1:1000) {
  perm_diffs[i] = perm_fun(session_times[, 'Time'], 21, 15)
\}
# Visualisasi Histogram & Proporsi p-value
hist(perm_diffs, xlab='Session time differences (in seconds)')
abline(v=obs_diff)
mean(perm_diffs > obs_diff)
```

# **11. Exhaustive and Bootstrap Permutation Tests**

### selain *Random Permutation Test*, terdapat varian:
  1. *Exhaustive Permutation Test*: Menghitung **seluruh** kemungkinan kombinasi pembagian kelompok (disebut juga *exact test*). Praktis hanya untuk sampel sangat kecil.
  2. *Bootstrap Permutation Test*: Pengambilan sampel saat pengocokan dilakukan dengan pengembalian (*with replacement*).

# **12. Permutation Tests: The Bottom Line for Data Science**

### Uji permutasi adalah alat heuristik yang sangat fleksibel (*one-size-fits-all*) untuk menguji signifikansi. Metode ini tidak membutuhkan asumsi distribusi normal, bebas dari rumus matematika yang kaku, serta mudah dijelaskan dan diimplementasikan pada data numerik maupun biner.

# **13. Statistical Significance and p-Values**

### Signifikansi statistik mengukur apakah hasil eksperimen berada di luar jangkauan variasi kebetulan acak. Jika hasil pengamatan sangat jarang terjadi di bawah hipotesis nol, hasil tersebut dinyatakan signifikan secara statistik.

# **14. p-Value**
### p-value adalah probabilitas mendapatkan hasil yang sama ekstremnya atau lebih ekstrem dari hasil observasi, jika Hipotesis Nol benar. p-value yang kecil (misal $\le 0.05$) menunjukkan bahwa hasil observasi jarang terjadi karena kebetulan semata.

**Code Repodruction: R:**

```python
# Uji proporsi biner A/B test (Approximation)
prop.test(x=c(200, 182), n=c(23739, 22588), alternative='greater')
```

# **15. Alpha**
### Alpha ($\alpha$) adalah ambang batas probabilitas tingkat signifikansi yang ditetapkan sebelum eksperimen dimulai (biasanya 5% atau 1%). Jika p-value < $\alpha$, Hipotesis Nol ditolak.

# **16. Type 1 and Type 2 Errors**

* **Type 1 Error:** Secara keliru menyimpulkan ada efek nyata, padahal sebenarnya hanya faktor kebetulan (menolak $H_0$ yang benar).
* **Type 2 Error:** Secara keliru menyimpulkan bahwa efek disebabkan oleh kebetulan, padahal sebenarnya efek tersebut nyata (gagal menolak $H_0$ yang salah).

# **17. Data Science and p-Values**
### *data scientist*, *p-value* bukanlah tujuan akhir melainkan sekadar instrumen filter untuk menilai apakah suatu pola/fitur layak diselidiki lebih lanjut. Dalam aplikasi bisnis, keberlanjutan dan besarnya dampak (*effect size*) jauh lebih penting daripada nilai *p-value* yang sangat kecil pada dataset besar.

# **18. t-Tests**

### Uji t (*t-test*) dikembangkan oleh William Sealy Gosset ("Student") sebagai metode parametrik standar untuk membandingkan rata-rata dua kelompok numerik tanpa harus melakukan *resampling* ribuan kali.

**Code Reproduction: R:**

```python
t.test(Time ~ Page, data=session_times, alternative='less')$0
```

# **19. Multiple Testing**
### Semakin banyak pengujian hipotesis yang dilakukan pada data yang sama, semakin tinggi risiko terjadinya Type 1 Error secara tidak sengaja (dikenal sebagai Alpha Inflation). Jika menguji 20 variabel acak pada $\alpha = 0.05$, peluang setidaknya satu variabel terlihat "signifikan secara semu" adalah $1 - (0.95)^{20} \approx 64\%$. Teknik penyesuaian seperti Bonferroni Correction atau pengontrolan False Discovery Rate (FDR) digunakan untuk mengatasi masalah ini.

# **20. Degrees of Freedom**

### Derajat kebebasan (*Degrees of Freedom / df*) merujuk pada jumlah nilai dalam perhitungan statistik akhir yang bebas bervariasi. Misalnya, jika 10 nilai memiliki rata-rata tertentu, hanya 9 nilai yang bebas bervariasi karena nilai ke-10 sudah ditentukan secara pasti.

# **21. ANOVA (Analysis of Variance)**
### ANOVA adalah prosedur statistik untuk membandingkan rata-rata dari **banyak kelompok** (misal A/B/C/D test) sekaligus dalam satu uji menyeluruh (*omnibus test*), tanpa harus melakukan banyak pengujian pasangan (*pairwise comparisons*).

**Code Reproduction: R:**

```python
# ANOVA berbasis Permutasi
library(lmPerm)
summary(aovp(Time ~ Page, data=four_sessions))

# ANOVA Klasik
summary(aov(Time ~ Page, data=four_sessions))
```

# **22. F-Statistic**
### Statistik F (F-statistic) dihitung dari rasio variansi antar rata-rata kelompok (between-group variance) terhadap variansi di dalam tiap kelompok (within-group variance / residual error). Semakin tinggi nilai F, semakin besar kemungkinan bahwa perbedaan antar kelompok bukan disebabkan oleh kebetulan acak.

# **23. Two-Way ANOVA**
### Two-Way ANOVA memperluas analisis dengan mengevaluasi pengaruh dua faktor kategorikal secara bersamaan serta mengukur efek interaksi (interaction effect) antara kedua faktor tersebut terhadap variabel respon kontinu.

# **24. Chi-Square Test **
### Uji Chi-Square ($\chi^2$) digunakan untuk data frekuensi/hitungan pada variabel kategorikal guna menguji seberapa besar simpangan frekuensi observasi dari frekuensi ekspektasi di bawah asumsi independensi (Hipotesis Nol).

# **25. Chi-Square Test: A Resampling Approach**

### Menguji perbedaan frekuensi antar kelompok kategorikal dengan pendekatan *resampling* acak untuk membuat distribusi Chi-Square sintetis.

**Code Reproduction: R:**

```python
# Chi-Square dengan Simulasi Resampling
chisq.test(clicks, simulate.p.value=TRUE)$0
```

# **26. Chi-Square Test: Statistical Theory**
### Menurut teori statistik asimptotik, distribusi statistik $\chi^2$ dapat diapproksimasi dengan distribusi Chi-Square teoritis berdasarkan derajat kebebasannya: $df = (r - 1) \times (c - 1)$, di mana $r$ adalah jumlah baris dan $c$ adalah jumlah kolom.

**Code Reproduction: R:**

```python
chisq.test(clicks, simulate.p.value=FALSE)
```

# **27. Fisher’s Exact Test**
### Ketika jumlah sampel/hitungan dalam sel sangat kecil (misalnya $\le 5$), aproksimasi distribusi Chi-Square tidak lagi akurat. Fisher's Exact Test mengelaborasi seluruh kemungkinan kombinasi pengurutan secara tepat untuk mendapatkan p-value yang presisi.

**Code Reproduction: R:**

```python
fisher.test(clicks)
```

# **30. Power and Sample Size**
### Sebelum menjalankan eksperimen web, peneliti harus menentukan berapa banyak ukuran sampel yang dibutuhkan. Terdapat 4 komponen yang saling terkait:
1. Sample Size (Ukuran sampel $n$)
2. Effect Size (Ukuran efek minimum yang ingin dideteksi, misal kenaikan konversi 10%)
3. Significance Level (Alpha $\alpha$)
4. Power (Probabilitas berhasil mendeteksi effect size jika efek tersebut benar-benar ada, biasanya diset 80%).

# **31. Sample Size**
### Menghitung ukuran sampel yang diperlukan berdasarkan *Effect Size*, *Alpha*, dan *Power* yang ditentukan.

**Code reproduction: R:**

```python
library(pwr)

# Menghitung sampel size untuk uji beda dua proporsi
h_effect <- pwr.2p.test(p1=0.0121, p2=0.011)\$h
pwr.2p.test(h=h_effect, sig.level=0.05, power=0.8, alternative='greater')
```
