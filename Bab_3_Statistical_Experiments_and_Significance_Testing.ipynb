# **Ringkasan**
### Desain eksperimen adalah fondasi utama pengujian statistik untuk mengonfirmasi atau menolak suatu hipotesis. Di dunia *data science*, eksperimen dilakukan secara terus menerus (seperti uji tampilan antarmuka web dan strategi pemasaran) untuk menguji efektivitas perlakuan (*treatment*) baru dibandingkan dengan kondisi standar.


# **1. A/B Testing**

### Uji A/B (*A/B Test*) adalah eksperimen yang membandingkan dua kelompok (*treatment A vs treatment B*) untuk menentukan perlakuan mana yang memberikan hasil lebih baik. Biasanya, salah satu kelompok berfungsi sebagai **Control Group** (perlakuan standar/tanpa perlakuan) dan kelompok lainnya sebagai **Treatment Group**. Subjek (misal: pengunjung web) dialokasikan secara acak (*randomization*) ke dalam kelompok A atau B, dan efeknya diukur menggunakan instrumen statistik tunggal (*test statistic*).

# **2. Why Have a Control Group?**
###Tanpa kelompok kontrol (*control group*), tidak ada jaminan bahwa perubahan hasil benar-benar dipicu oleh perlakuan (*treatment*) dan bukan oleh faktor ekstrim/luar lainnya (seperti perubahan musim, waktu, atau karakteristik pengguna). Kelompok kontrol memastikan kondisi *ceteris paribus* (semua hal lain dianggap sama).

# **3. Why Just A/B? Why Not C, D,...?**

###Eksperimen web tidak terbatas pada dua opsi. Peneliti dapat menguji banyak variasi sekaligus (A/B/C/D). Namun, semakin banyak variasi yang diuji secara bersamaan dalam statistik tradisional, semakin rumit pengujian signifikansinya. Dalam *data science*, fokus utama sering kali beralih dari sekadar menjawab "apakah A berbeda dari B?" menjadi "mana opsi terbaik dari banyak pilihan?" yang nantinya diselesaikan dengan algoritma *Multi-Arm Bandit*.

# **4. Hypothesis Tests**

### Pengujian hipotesis (*hypothesis testing*) diciptakan untuk melindungi peneliti agar tidak terkecoh oleh variasi acak (*random chance*). Otak manusia cenderung mencari pola dan menganggap perbedaan acak kecil sebagai sesuatu yang signifikan secara nyata.

# **5. The Null HypothesisRingkasan:
### Hipotesis Nol ($H_0$) adalah andaikan dasar bahwa tidak ada perbedaan nyata antara kelompok A dan B, serta perbedaan yang teramati semata-mata disebabkan oleh kebetulan/variasi acak. Tujuan eksperimen adalah mengumpulkan bukti untuk menolak atau meragukan Hipotesis Nol ini.**

# **6. Alternative HypothesisRingkasan:**
### Hipotesis Alternatif ($H_a$) adalah lawan dari Hipotesis Nol. Jika $H_0$ menyatakan "A = B", maka $H_a$ menyatakan "A ≠ B" (atau "B lebih baik dari A").

# **7. One-Way Versus Two-Way Hypothesis Tests**

### Pengujian satu arah (*one-tail / one-way test*) hanya memperhitungkan arah perbedaan yang diharapkan (misal: B &gt; A). Pengujian dua arah (*two-tail / two-way test*) lebih konservatif karena memperhitungkan kemungkinan perbedaan ekstrem di kedua arah (A &gt; B atau B &gt; A).

# **8. Resampling**
### *Resampling* adalah proses pengambilan sampel berulang dari data hasil observasi. Dua teknik utama *resampling* adalah *bootstrapping* (untuk mengukur presisi estimasi) dan *permutation test* (untuk menguji hipotesis antar kelompok).

# **9. Permutation Test**

### Dalam uji permutasi (*permutation test* atau *randomization test*), data dari seluruh kelompok digabungkan menjadi satu *pool* (mencerminkan Hipotesis Nol bahwa semua perlakuan sama), kemudian dikocok (*shuffled*) dan dibagi kembali secara acak menjadi kelompok-kelompok baru dengan ukuran yang sama dengan sampel asli. Selisih antar kelompok acak ini dihitung berulang kali untuk membentuk distribusi acak Hipotesis Nol.

# **10. Example: Web StickinessRingkasan:**
### Studi kasus membandingkan waktu sesi (session time) pengguna pada Halaman A (21 sampel) dan Halaman B (15 sampel). Rata-rata waktu sesi Halaman B lebih lama 35.67 detik dibanding Halaman A. Uji permutasi dijalankan sebanyak 1.000 kali untuk melihat seberapa sering kocokan acak menghasilkan perbedaan $\ge 35.67$ detik.

**Code Repodruction: R:**

```python
library(ggplot2)

# 1. Boxplot Visualisasi
ggplot(session_times, aes(x=Page, y=Time)) + geom_boxplot()
# 2. Perhitungan Selisih Mean Observasi
mean_a <- mean(session_times[session_times['Page'] == 'Page A', 'Time'])
mean_b <- mean(session_times[session_times['Page'] == 'Page B', 'Time'])
obs_diff <- mean_b - mean_a
# 3. Fungsi Permutasi Kustom
perm_fun <- function(x, nA, nB) {
  n <- nA + nB
  idx_b <- sample(1:n, nB)
  idx_a <- setdiff(1:n, idx_b)
  return(mean(x[idx_b]) - mean(x[idx_a]))
\}
# 4. Simulasi 1.000 kali
perm_diffs <- rep(0, 1000)
for (i in 1:1000) {
  perm_diffs[i] = perm_fun(session_times[, 'Time'], 21, 15)
\}
# Visualisasi Histogram & Proporsi p-value
hist(perm_diffs, xlab='Session time differences (in seconds)')
abline(v=obs_diff)
mean(perm_diffs > obs_diff)
```

# **11. Exhaustive and Bootstrap Permutation Tests**

### selain *Random Permutation Test*, terdapat varian:
  1. *Exhaustive Permutation Test*: Menghitung **seluruh** kemungkinan kombinasi pembagian kelompok (disebut juga *exact test*). Praktis hanya untuk sampel sangat kecil.
  2. *Bootstrap Permutation Test*: Pengambilan sampel saat pengocokan dilakukan dengan pengembalian (*with replacement*).

# **12. Permutation Tests: The Bottom Line for Data Science**

### Uji permutasi adalah alat heuristik yang sangat fleksibel (*one-size-fits-all*) untuk menguji signifikansi. Metode ini tidak membutuhkan asumsi distribusi normal, bebas dari rumus matematika yang kaku, serta mudah dijelaskan dan diimplementasikan pada data numerik maupun biner.

# **13. Statistical Significance and p-Values**

### Signifikansi statistik mengukur apakah hasil eksperimen berada di luar jangkauan variasi kebetulan acak. Jika hasil pengamatan sangat jarang terjadi di bawah hipotesis nol, hasil tersebut dinyatakan signifikan secara statistik.
