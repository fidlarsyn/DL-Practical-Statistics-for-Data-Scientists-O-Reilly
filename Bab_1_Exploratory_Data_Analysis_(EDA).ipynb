## **1. Elements of Structured Data**
###Data mentah mentah (seperti teks, gambar, atau *clickstream*) harus diolah menjadi bentuk terstruktur agar dapat dianalisis secara statistik. Data terstruktur diklasifikasikan ke dalam tipe data utama: **Numeric** (terdiri dari *continuous* seperti tinggi badan, dan *discrete* seperti jumlah hitungan) serta **Categorical** (terdiri dari *binary*/indikator 0/1, dan *ordinal* yang memiliki urutan eksplisit). Klasifikasi tipe data menentukan metode visualisasi dan komputasi yang digunakan oleh perangkat lunak seperti R dan Python

# **2. Rectangular Data**
###Format standar untuk analisis *data science* adalah *rectangular data* (tabel dua dimensi). Baris mewakili **records** (kasus/observasi), dan kolom mewakili **features** (variabel/prediktor). Dalam pemodelan, satu atau lebih *features* digunakan untuk memprediksi **outcome** (target/variabel dependen).

# **3. Data Frames and Indexes**
### *Data frame* adalah struktur *rectangular data* utama dalam R (`data.frame`) dan Python (`pandas.DataFrame`). Di Python `pandas`, *DataFrame* secara bawaan membuat indeks integer otomatis untuk urutan baris dan mendukung indeks hierarki/multilevel. Di R, paket seperti `data.table` dan `dplyr` mempercepat manipulasi *data frame*.

# **4. Nonrectangular Data Structures**
### Selain data persegi (*rectangular*), terdapat struktur data lain seperti **Time Series** (pengukuran berurutan sepanjang waktu), **Spatial Data** (data pemetaan berbasis koordinat lokasi atau piksel), dan **Graph/Network Data** (menggambarkan hubungan antar entitas seperti jejaring sosial atau rute distribusi).

# **5. Estimates of Location**
### Mengukur estimasi nilai pusat/tipikal (*central tendency*) dari suatu fitur numerik.
* **Mean**: Rata-rata aritmatika dari semua nilai data (jumlah nilai dibagi banyaknya data). Sangat sensitif terhadap nilai ekstrem (*outliers*).
* **Median and Robust Estimates**: *Median* adalah nilai tengah dari data terurut. *Median* bersifat *robust* (tidak sensitif terhadap *outliers*)|. *Trimmed mean* (rata-rata terpotong) membuang persentase tertentu dari nilai ekstrem di ujung atas dan bawah sebagai kompromi antara *mean* dan *median*. *Weighted mean* dan *weighted median* memperhitungkan bobot individual tiap observasi.
* **Example: Location Estimates of Population and Murder Rates**: Menguji perhitungan estimasi lokasi menggunakan data populasi dan tingkat pembunuhan (*murder rate*) di negara bagian AS.

**Code Reproduction: R:**

```python
state &lt;- read.csv('state.csv')
# Mean, Trimmed Mean (memotong 10% di setiap ujung), Median
mean(state[['Population']])
mean(state[['Population']], trim=0.1)
median(state[['Population']])

# Weighted Mean &amp; Weighted Median
weighted.mean(state[['Murder.Rate']],
              w=state[['Population']])
library('matrixStats')
weighted.median(state[['Murder.Rate']],
              w=state[['Population']])
```

# **6. Estimates of Variability**
### Mengukur penyebaran (*dispersion*) data, yaitu seberapa rapat atau tersebarnya nilai-nilai data di sekitar pusat.
* **Standard Deviation and Related Estimates:**  Berbasis deviasi (selisih nilai dari rata-rata)22. Variance adalah rata-rata kuadrat deviasi, sedangkan Standard Deviation adalah akar kuadrat dari variansi22. Mean Absolute Deviation (MAD) mengambil rata-rata nilai absolut deviasi22. Sebagai alternatif robust, Median Absolute Deviation from the median (MAD) mengukur median dari selisih absolut terhadap median data23.
* **Estimates Based on Percentiles:** Mengukur penyebaran berdasarkan data yang terurut (order statistics)24. Range adalah selisih nilai maksimum dan minimum2425. Interquartile Range (IQR) adalah selisih antara persentil ke-75 ($Q_3$) dan persentil ke-25 ($Q_1$), tidak sensitif terhadap outlier2526.
* **Example:** Variability Estimates of State Population: Menghitung ukuran variabilitas pada data populasi negara bagian

**Code Reproduction: R:**

```python
# Standard Deviation, IQR, MAD
sd(state[['Population']])
mad(state[['Population']])
iqr(state[['Population']])
```

# **7. Exploring the Data Distribution**
### Mengeksplorasi bentuk keseluruhan distribusi data.
* **Percentiles and Boxplots:** Boxplot (diagram kotak-garis yang diperkenalkan oleh Tukey) mengvisualisasikan distribusi data secara cepat berdasarkan kuartil ($Q_1$, Median, $Q_3$) serta whiskers (hingga $1.5 \times \text{IQR}$) dan outliers.
* **Frequency Tables and Histograms:** Frequency Table membagi rentang data numerik menjadi interval (bins) berukuran sama dan menghitung jumlah observasi dalam tiap bin3033. Histogram menggambarkan frequency table dalam bentuk diagram batang berdampingan.
* **Density Plots and Estimates:** Density plot adalah versi halus (smooth) dari histogram berdasarkan Kernel Density Estimate (KDE), di mana total area di bawah kurva sama dengan 1.

**Code Reproduction: R:**

```python
# Quantiles &amp; Boxplot
quantile(state[['Murder.Rate']], p=c(.05, .25, .5, .75, .95))
boxplot(state[['Population']]/1000000, ylab='Population (millions)')

# Frequency Table &amp; Histogram
breaks &lt;- seq(from=min(state[['Population']]),
                 to=max(state[['Population']]), length=11)
pop_freq &lt;- cut(state[['Population']], breaks=breaks,
                   right=TRUE, include.lowest=TRUE)
table(pop_freq)
hist(state[['Population']], breaks=breaks)

# Density Plot
hist(state[['Murder.Rate']], freq=FALSE)
lines(density(state[['Murder.Rate']]), lwd=3, col='blue')
```
